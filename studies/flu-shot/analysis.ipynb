## 1. 후보 대회 1: Flu Shot Learning: Predict H1N1 and Seasonal Flu Vaccines

### 1.1 참가 대회명 및 대회 기본 정보

- 대회명: Flu Shot Learning: Predict H1N1 and Seasonal Flu Vaccines
- 플랫폼: DrivenData
- 문제 유형: 다중 라벨 이진 분류 문제
- 예측 대상: `h1n1_vaccine`, `seasonal_vaccine`
- 평가 지표: ROC AUC
- 대회 성격: Beginner Practice Competition
- 기간: DrivenData 연습 대회로 현재도 참가 및 제출 가능
- 상금: 별도 상금 없음
- 제출 파일 형식: `respondent_id`, `h1n1_vaccine`, `seasonal_vaccine`

Flu Shot Learning은 응답자의 배경, 건강 행동, 질병과 백신에 대한 인식 정보를 이용하여 H1N1 백신과 계절성 독감 백신 접종 여부를 예측하는 대회이다.

이 대회는 하나의 정답만 예측하는 일반적인 이진 분류 문제가 아니라, 두 개의 target을 동시에 예측하는 다중 라벨 분류 문제이다. 예측 대상은 `h1n1_vaccine`과 `seasonal_vaccine`이며, 각각 H1N1 백신 접종 여부와 계절성 독감 백신 접종 여부를 의미한다.

최종 제출에서는 단순히 0 또는 1을 제출하는 것이 아니라, 각 응답자가 해당 백신을 접종했을 확률을 예측하여 제출한다.

### 1.2 대회 선정 이유

Flu Shot Learning을 후보 대회로 선정한 이유는 백신 접종 여부가 단순한 의학적 문제만이 아니라 개인의 인식, 행동, 사회적 배경과도 관련된 문제이기 때문이다.

백신 접종은 감염병 예방에서 중요한 역할을 한다. 하지만 모든 사람이 같은 조건에서 백신을 접종하는 것은 아니다. 질병에 대한 위험 인식, 백신 효과에 대한 신뢰, 의사의 권고 여부, 건강보험 여부, 직업, 연령, 교육 수준 등 여러 요인이 접종 여부에 영향을 줄 수 있다.

이 대회는 설문 데이터를 기반으로 하므로 결측치 처리, 범주형 변수 인코딩, 순서형 변수 처리, 다중 target 예측 등 다양한 데이터 분석 과정을 포함한다. 따라서 데이터공학 과목에서 다루는 데이터 전처리와 모델링 과정을 적용하기에 적합하다.

또한 공중보건과 연결되는 주제이므로 발표에서도 의미 있게 다룰 수 있다. 어떤 특성을 가진 사람들이 백신을 더 많이 접종했는지 분석하면 단순한 예측을 넘어서 정책적 시사점도 설명할 수 있다.

### 1.3 대회 목적, 평가 지표 및 대회 규칙

이 대회의 목적은 설문 응답 데이터를 이용하여 개인이 H1N1 백신과 계절성 독감 백신을 접종했을 가능성을 예측하는 것이다.

예측 대상은 다음 두 가지이다.

- `h1n1_vaccine`: H1N1 백신 접종 여부
- `seasonal_vaccine`: 계절성 독감 백신 접종 여부

평가 지표는 ROC AUC이다. ROC AUC는 모델이 양성 클래스와 음성 클래스를 얼마나 잘 구분하는지 평가하는 지표이다. 이 대회에서는 두 target에 대해 각각 ROC AUC를 계산하고, 두 점수의 평균을 최종 점수로 사용한다.

ROC AUC는 1에 가까울수록 좋은 성능을 의미한다. 0.5는 무작위 예측과 비슷한 수준이고, 1.0은 완벽하게 구분한 경우이다.

대회 규칙은 학습용 feature 데이터와 label 데이터를 이용하여 모델을 학습하고, test 데이터에 대해 두 백신 접종 확률을 예측하는 방식이다. 제출 파일은 `submission_format.csv`와 동일한 구조를 가져야 하며, `respondent_id`, `h1n1_vaccine`, `seasonal_vaccine` 컬럼을 포함해야 한다.

### 1.4 데이터셋 정보

Flu Shot Learning 데이터셋은 다음 파일로 구성된다.

- `training_set_features.csv`: 학습용 입력 변수
- `training_set_labels.csv`: 학습용 정답 변수
- `test_set_features.csv`: 평가용 입력 변수
- `submission_format.csv`: 제출 파일 형식

`training_set_features.csv`에는 응답자의 설문 응답 정보가 포함되어 있다. 변수들은 질병에 대한 걱정 정도, 백신 효과에 대한 인식, 의사의 권고 여부, 건강 행동, 직업, 연령대, 교육 수준, 소득 수준 등으로 구성된다.

`training_set_labels.csv`에는 두 개의 정답 컬럼이 포함된다.

- `h1n1_vaccine`
- `seasonal_vaccine`

`test_set_features.csv`에는 정답이 포함되어 있지 않으며, 이 데이터에 대해 두 백신 접종 확률을 예측해야 한다. `submission_format.csv`는 제출 파일의 기본 형식을 제공한다.

```python
# Flu Shot Learning 데이터 경로 설정
# 1순위: 현재 폴더
# 2순위: ./data/flu-shot 폴더
# 3순위: /mnt/data 폴더

possible_dirs = [".", "./data/flu-shot", "/mnt/data"]

DATA_DIR = None

for path in possible_dirs:
    if os.path.exists(os.path.join(path, "training_set_features.csv")):
        DATA_DIR = path
        break

if DATA_DIR is None:
    raise FileNotFoundError("training_set_features.csv 파일을 찾을 수 없습니다. 파일 위치를 확인하세요.")

features_path = os.path.join(DATA_DIR, "training_set_features.csv")
labels_path = os.path.join(DATA_DIR, "training_set_labels.csv")
test_path = os.path.join(DATA_DIR, "test_set_features.csv")
submission_path = os.path.join(DATA_DIR, "submission_format.csv")

flu_features = pd.read_csv(features_path)
flu_labels = pd.read_csv(labels_path)
flu_test = pd.read_csv(test_path)
flu_submission = pd.read_csv(submission_path)

print("데이터 경로:", DATA_DIR)
print("features shape:", flu_features.shape)
print("labels shape:", flu_labels.shape)
print("test shape:", flu_test.shape)
print("submission shape:", flu_submission.shape)
```

목표 변수 분포를 확인한 결과, H1N1 백신 접종자는 전체의 약 21.2%이고, 계절성 독감 백신 접종자는 전체의 약 46.6%이다.

즉, H1N1 백신은 접종하지 않은 사람이 훨씬 많고, 계절성 독감 백신은 접종자와 비접종자의 비율 차이가 상대적으로 작다. 두 target의 분포가 다르기 때문에 각각의 예측 난이도와 모델 성능도 다르게 나타날 수 있다.

이 대회에서는 두 target을 동시에 예측하지만, 각 target의 특성이 다르므로 `h1n1_vaccine`과 `seasonal_vaccine`을 따로 분석하는 것이 필요하다.

```python
pd.crosstab(
    flu_labels["h1n1_vaccine"],
    flu_labels["seasonal_vaccine"],
    normalize="all"
)
```

```python
pd.crosstab(
    flu_labels["h1n1_vaccine"],
    flu_labels["seasonal_vaccine"],
    margins=True
)
```

두 백신 접종 여부를 교차표로 확인하면, H1N1 백신과 계절성 독감 백신을 모두 접종한 응답자는 전체의 약 17.6%이다. 반대로 두 백신을 모두 접종하지 않은 응답자는 약 49.8%로 가장 큰 비중을 차지한다.

H1N1 백신은 접종하지 않았지만 계절성 독감 백신은 접종한 응답자도 약 29.0% 존재한다. 이는 두 백신 접종 여부가 완전히 같은 패턴으로 결정되는 것이 아니라는 점을 보여준다.

따라서 두 target은 서로 관련이 있지만 동일한 변수로 취급할 수는 없으며, 각각에 대해 별도의 예측 확률을 생성해야 한다.

```python
missing_train = flu_features.isnull().sum().sort_values(ascending=False)
missing_test = flu_test.isnull().sum().sort_values(ascending=False)

print("학습 데이터 결측치 상위 20개")
display(missing_train.head(20))

print("평가 데이터 결측치 상위 20개")
display(missing_test.head(20))
```

```python
missing_ratio = (flu_features.isnull().mean() * 100).sort_values(ascending=False)

plt.figure(figsize=(10, 6))
missing_ratio.head(15).sort_values().plot(kind="barh")
plt.title("Top 15 Missing Value Ratio")
plt.xlabel("Missing Ratio (%)")
plt.ylabel("Columns")
plt.show()
```

결측치 분석 결과, `employment_occupation`, `employment_industry`, `health_insurance`, `income_poverty` 변수에서 결측치가 많이 나타났다.

특히 `employment_occupation`과 `employment_industry`는 결측치가 약 절반 수준으로 많다. 이는 응답자가 고용 상태가 아니거나 직업 관련 문항에 응답하지 않았기 때문일 수 있다. `health_insurance`와 `income_poverty`도 결측치가 많아 모델링 전에 적절한 처리가 필요하다.

결측치가 적은 변수는 중앙값이나 최빈값으로 대체할 수 있지만, 결측치가 많은 변수는 단순 대체만으로는 정보가 왜곡될 수 있다. 따라서 결측 자체가 의미를 가질 수 있는 변수는 `Unknown` 또는 `Missing` 범주로 처리하는 방법도 고려할 수 있다.

### 1.6 주요 변수와 백신 접종률 관계 분석

백신 접종 여부와 관련성이 높을 것으로 예상되는 변수들을 중심으로 평균 접종률을 확인한다. 특히 의사의 백신 권고 여부, 의료 종사자 여부, 건강보험 여부, 연령대 등은 백신 접종과 관련이 있을 가능성이 높다.

```python
# 의사의 H1N1 백신 권고 여부에 따른 접종률

flu_train.groupby("doctor_recc_h1n1")[["h1n1_vaccine", "seasonal_vaccine"]].mean()
```

```python
flu_train.groupby("doctor_recc_h1n1")[["h1n1_vaccine", "seasonal_vaccine"]].mean().plot(kind="bar", figsize=(7, 4))
plt.title("Vaccination Rate by Doctor Recommendation for H1N1")
plt.xlabel("doctor_recc_h1n1")
plt.ylabel("Vaccination Rate")
plt.show()
```

의사의 H1N1 백신 권고 여부에 따라 H1N1 백신 접종률 차이가 크게 나타난다. 의사의 권고를 받지 않은 집단의 H1N1 백신 접종률은 약 13.6%이고, 권고를 받은 집단의 접종률은 약 53.2%이다.

이는 의사의 권고 여부가 H1N1 백신 접종을 예측하는 데 중요한 변수로 작용할 수 있음을 보여준다.

```python
# 의사의 계절성 독감 백신 권고 여부에 따른 접종률

flu_train.groupby("doctor_recc_seasonal")[["h1n1_vaccine", "seasonal_vaccine"]].mean()
```

```python
flu_train.groupby("doctor_recc_seasonal")[["h1n1_vaccine", "seasonal_vaccine"]].mean().plot(kind="bar", figsize=(7, 4))
plt.title("Vaccination Rate by Doctor Recommendation for Seasonal Vaccine")
plt.xlabel("doctor_recc_seasonal")
plt.ylabel("Vaccination Rate")
plt.show()
```

계절성 독감 백신도 의사의 권고 여부에 따라 접종률 차이가 크게 나타난다. 의사의 계절성 독감 백신 권고를 받지 않은 집단의 계절성 독감 백신 접종률은 약 34.6%이고, 권고를 받은 집단의 접종률은 약 73.8%이다.

따라서 의사의 권고 여부는 두 백신 접종 여부 모두와 강한 관련성을 가지는 변수로 볼 수 있다.

```python
age_vaccine_rate = flu_train.groupby("age_group")[["h1n1_vaccine", "seasonal_vaccine"]].mean()
age_vaccine_rate
```

```python
age_order = ["18 - 34 Years", "35 - 44 Years", "45 - 54 Years", "55 - 64 Years", "65+ Years"]

age_vaccine_rate = flu_train.groupby("age_group")[["h1n1_vaccine", "seasonal_vaccine"]].mean()
age_vaccine_rate = age_vaccine_rate.reindex(age_order)

age_vaccine_rate.plot(kind="bar", figsize=(9, 5))
plt.title("Vaccination Rate by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Vaccination Rate")
plt.xticks(rotation=45)
plt.show()
```

연령대별 분석 결과, 계절성 독감 백신 접종률은 나이가 많을수록 높아지는 경향을 보인다. 특히 65세 이상 집단의 계절성 독감 백신 접종률은 약 67.4%로 가장 높다.

반면 H1N1 백신 접종률은 연령대에 따라 차이가 있긴 하지만 계절성 독감 백신만큼 뚜렷하게 증가하지는 않는다. 이는 H1N1 백신과 계절성 독감 백신의 접종 결정 요인이 일부 다를 수 있음을 보여준다.

```python
health_cols = ["health_worker", "health_insurance", "chronic_med_condition"]

for col in health_cols:
    if col in flu_train.columns:
        print(f"\n[{col}]")
        display(flu_train.groupby(col)[["h1n1_vaccine", "seasonal_vaccine"]].mean())
```

```python
for col in health_cols:
    if col in flu_train.columns:
        flu_train.groupby(col)[["h1n1_vaccine", "seasonal_vaccine"]].mean().plot(kind="bar", figsize=(7, 4))
        plt.title(f"Vaccination Rate by {col}")
        plt.xlabel(col)
        plt.ylabel("Vaccination Rate")
        plt.show()
```

건강 관련 변수에서도 백신 접종률 차이가 나타난다. 의료 종사자인 경우 H1N1 백신과 계절성 독감 백신 접종률이 모두 높게 나타난다. 건강보험이 있는 집단도 백신 접종률이 상대적으로 높게 나타난다.

이는 백신 접종이 개인의 건강 상태뿐만 아니라 의료 접근성과 직업적 특성과도 관련될 수 있음을 보여준다.

```python
socio_cols = ["education", "income_poverty", "sex", "race"]

for col in socio_cols:
    if col in flu_train.columns:
        print(f"\n[{col}]")
        display(flu_train.groupby(col)[["h1n1_vaccine", "seasonal_vaccine"]].mean())
```

```python
for col in ["education", "income_poverty"]:
    if col in flu_train.columns:
        flu_train.groupby(col)[["h1n1_vaccine", "seasonal_vaccine"]].mean().plot(kind="bar", figsize=(9, 5))
        plt.title(f"Vaccination Rate by {col}")
        plt.xlabel(col)
        plt.ylabel("Vaccination Rate")
        plt.xticks(rotation=45)
        plt.show()
```

교육 수준과 소득 수준에 따라 백신 접종률 차이가 나타난다. 대학 졸업 집단은 상대적으로 백신 접종률이 높고, 소득이 높은 집단도 백신 접종률이 높은 경향을 보인다.

이는 백신 접종 여부가 질병 인식뿐만 아니라 사회경제적 요인과도 관련될 수 있음을 의미한다.

### 1.7 진행 계획

Flu Shot Learning 대회는 이 순서로 진행할 계획이다.

1. 데이터 파일 구조 확인
2. 학습 feature와 label 데이터 병합
3. 두 target 변수 분포 확인
   - `h1n1_vaccine`
   - `seasonal_vaccine`
4. 결측치 비율 확인
5. 수치형 변수와 범주형 변수 구분
6. 주요 변수와 백신 접종률 관계 분석
   - 의사의 백신 권고 여부
   - 연령대
   - 건강보험 여부
   - 의료 종사자 여부
   - 교육 수준
   - 소득 수준
7. 결측치 처리
   - 수치형 변수: 중앙값 대체
   - 범주형 변수: 최빈값 또는 `Unknown` 대체
8. 범주형 변수 인코딩
   - One-Hot Encoding 적용
9. 모델 학습
   - Logistic Regression
   - Random Forest
   - Gradient Boosting
   - LightGBM
10. ROC AUC 기준 성능 비교
11. 두 target에 대한 예측 확률 생성
12. 제출 파일 작성
